# Unix Shell Tutorial: XML Data Handling

This is the **fifth tutorial** in a series that will demonstrate how shell scripting can be used to perform the tasks that health and life science specialists may need to undertake to find and retrieve biomedical data and text. We will use the compound caffeine as an example and explore different public repositories to identify diseases related to it. The focus is not on the specific relationships we may discover, but on the process of obtaining them.

The objective of this tutorial is to learn XML data handling techniques using shell scripting, in order to search, filter, and extract information from the XML files, which in our case contain information characterizing a protein.

> This tutorial is part of a series of tutorials adapted as interactive versions of the hands-on steps described in the [Data and Text Processing for Health and Life Sciences](https://labs.rd.ciencias.ulisboa.pt/book/) book, which is licensed under the [CC BY 4.0](https://creativecommons.org/licenses/by/4.0/).

## Step 1: Filtering Human Proteins

Assuming that our information need only concerns human diseases, we have to process the XML file of each protein to check if it represents a Homo sapiens (Human) protein.

To get started, we first need to retrieve the data file generated in the previous tutorial. The following command downloads the chebi_27732_xrefs_UniProt.csv file directly from the GitHub repository:

````bash
%%bash
curl -s -O 'https://raw.githubusercontent.com/lasigeBioTM/data-text-processing-notebooks/refs/heads/main/data/chebi_27732_proteins_xml.zip'
unzip -o chebi_27732_proteins_xml.zip
````

### Human proteins

To perform this filter, we can again use the `grep` command, to select only the lines of any XML file that specify the organism as Homo sapiens:

````bash
%%bash
grep '<name type="scientific">Homo sapiens</name>' chebi_27732_*.xml
````

**Expected Output:** Display of filenames that represent a human protein, with lines showing the Homo sapiens organism specification.

The output displays the filenames that represent a human protein.

Note that since the asterisk character (`*`) provides multiple files as arguments to `grep`, the ones whose name starts with `chebi_27732_` and ends with `.xml`, the output now includes the filename (followed by a colon (`:`)) where each line was matched.

We can use the `cut` command to extract only the filename, but grep has the `-l` option to just print the filename:

````bash
%%bash
grep -l '<name type="scientific">Homo sapiens</name>' chebi_27732_*.xml
````

**Expected Output:** List of filenames representing human proteins (e.g., chebi_27732_P21817.xml, chebi_27732_P12345.xml, etc.)

The equivalent long form to the `-l` option is `--files-with-matches`.

The output will now show only the filenames. These eight files represent the eight human proteins related to caffeine.

### PubMed identifiers

Now we need to extract the PubMed identifiers from these files to retrieve the related publications. For example, if we execute the following command:

````bash
%%bash
grep '<dbReference type="PubMed"' chebi_27732_P21817.xml
````

**Expected Output:** A long list of lines showing PubMed database references with their attributes from the P21817 protein file.

The output is a long list of publications related to protein P21817.

In the next step, we will learn how to extract these identifiers from all human proteins.

## Step 2: Extracting and Deduplicating PubMed Identifiers

To extract just the identifier, we can again use the cut command:

````bash
%%bash
grep '<dbReference type="PubMed"' chebi_27732_P21817.xml | cut -d'"' -f4
````

**Expected Output:** List of PubMed identifiers extracted from the dbReference elements.

We should note that `"` is used as the separation character and, since the PubMed identifier appears after the third `"`, the `4` represents the identifier. Now the output should be just the list of identifiers.

````bash
%%bash
grep '<name type="scientific">Homo sapiens</name>' chebi_27732_*.xml
````

**Expected Output:** List of lines showing Homo sapiens organism specifications from multiple XML files.

### PubMed identifiers extraction

Now to apply to every protein we may again use the `xargs` command:

````bash
%%bash
grep -l '<name type="scientific">Homo sapiens</name>' chebi_27732_*.xml | \
xargs -I {} grep '<dbReference type="PubMed"' {} | \
cut -d'"' -f4
````

**Expected Output:** Complete list of PubMed identifiers extracted from all human protein files, including duplicate entries.

This may provide a long list of PubMed identifiers, including repetitions since the same publication can be cited in different entries.

### Duplicate removal

To help us identify the repetitions, we can add the sort command, which will display the repeated identifiers in consecutive lines (due by sorting all identifiers):

````bash
%%bash
grep -l '<name type="scientific">Homo sapiens</name>' chebi_27732_*.xml | \
xargs -I {} grep '<dbReference type="PubMed"' {} | \
cut -d'"' -f4 | \
sort
````

**Expected Output:** Sorted list of PubMed identifiers, with duplicates appearing consecutively.

For example, some repeated PubMed identifiers that we should easily be able to see, such as `9607712`.

Fortunately, we also have the `-u` option that removes all these duplicates:

````bash
%%bash
grep -l '<name type="scientific">Homo sapiens</name>' chebi_27732_*.xml | \
xargs -I {} grep '<dbReference type="PubMed"' {} | \
cut -d'"' -f4 | \
sort -u
````

**Expected Output:** List of unique PubMed identifiers with all duplicates removed.

To easily check how many duplicates were removed, we can use the word count `wc` command with and without the usage of the `-u` option:

````bash
%%bash
grep -l '<name type="scientific">Homo sapiens</name>' chebi_27732_*.xml | \
xargs -I {} grep '<dbReference type="PubMed"' {} | \
cut -d'"' -f4 | \
sort | \
wc
````

**Expected Output:** Showing the total number of PubMed identifiers including duplicates.

````bash
%%bash
grep -l '<name type="scientific">Homo sapiens</name>' chebi_27732_*.xml | \
xargs -I {} grep '<dbReference type="PubMed"' {} | \
cut -d'"' -f4 | \
sort -u | \
wc
````

**Expected Output:** Showing the count of unique PubMed identifiers.

In case we have in our folder any auxiliary file, such as `chebi_27732_P21817_entry.xml`, we should add the option `--exclude *entry.xml` to the first `grep` command.

`wc` prints the numbers of lines, words, and bytes, thus in our case we are interested in first number. We can see that we have removed 605 − 276 = 329 duplicates.

Just for curiosity, we can also use the shell to perform simple mathematical calculations using the `expr` command:

````bash
%%bash
expr 605 - 276
````

**Expected Output:** 329 (the number of duplicates removed)

Now let us create a script file named getpublications.sh:

````bash
%%bash
cat > getpublications.sh << 'EOF'
# CHEBI identifier given as input is renamed to ID
ID=$1

grep -l '<name type="scientific">Homo sapiens</name>' chebi_$ID_*.xml | \
    xargs \
        -I {} grep '<dbReference type="PubMed"' {} | \
    cut -d'"' -f4 | \
    sort -u
EOF
````

Again, do not forget to save it in our working directory, and add the right permissions with chmod as we did previously with the other scripts.

````bash
%%bash
chmod u+x getpublications.sh
````

**Expected Output:** No output (permissions successfully changed).

To execute the script again:

````bash
%%bash
./getpublications.sh 27732
````

**Expected Output:** List of the unique PubMed identifiers related to caffeine in human proteins.

We can verify how many unique publications were obtained by using the `-l` option of `wc`, that provides only the number of lines:

````bash
%%bash
./getpublications.sh 27732 | wc -l
````

## Step 3: Advanced XML Parsing with XPath

### Complex Elements

Not always the XML elements are in the same line, as fortunately was the case of the PubMed identifiers. In those cases, we may have to use the `xmllint` command, a parser that is able to extract data through the specification of a XPath query, instead of using a single line pattern as in `grep`.

To install `xmllint` we can execute:

````bash
%%bash
apt-get update && apt-get install -y libxml2-utils
````

### XPath

XPath (XML Path Language) is a powerful tool to extract information from XML and HTML documents by following their hierarchical structure. Check W3C for more about XPath syntax.

### Namespace problems

In the case of our protein XML files, we can see that their second line defines a specific namespace using the `xmlns` attribute:

```xml
<uniprot
    xmlns="http://uniprot.org/uniprot"
    xmlns:xsi="http://www.w3.org/2001/XMLSchema-instance"
    xsi:schemaLocation="http://uniprot.org/uniprot
    http://www.uniprot.org/support/docs/uniprot.xsd">
```

This complicates our XPath queries, since we need to explicitly specify that we are using the local name for every element in a XPath query. For example, to get the data in each reference element:

````bash
%%bash
xmllint --nsclean --xpath "//*[local-name()='reference']" chebi_27732_P21817.xml
````

**Expected Output:** XML elements containing all reference information from the protein file.

Note that `//` means any path in the XML file until reaching a reference element. The square brackets in XPath queries normally represent conditions that need to be verified. The `--nsclean` removes the redundant namespace declaration in the `entry` element.

### Only local names

If we are only interested in using local names there is a way to avoid the usage of local-name() for every element in a XPath query. We can identify the top-level element, in our case entry, and extract all the data that it encloses using a XPath query. For example, we can create the auxiliary file `chebi_27732_P21817_entry.xml` by adding the redirection operator:

````bash
%%bash
xmllint --nsclean --xpath "//*[local-name()='entry']" chebi_27732_P21817.xml > chebi_27732_P21817_entry.xml
````

**Expected Output:** File created successfully (no output).

The new XML file now starts and ends with the `entry` element without any namespace definition:

```xml
<entry dataset="Swiss-Prot" created="1991-05-01" ...
<accession>P21817</accession>
...
</sequence>
</entry>
```

We can now apply any XPath query, for example `//reference`, on the auxiliary file without the need to explicitly say that it represents a local name:

````bash
%%bash
xmllint --xpath '//reference' chebi_27732_P21817_entry.xml
````

**Expected Output:** XML reference elements from the cleaned entry file.

The output contains only the data inside each `reference` element.

### Queries

The XPath syntax allows us to create many useful queries, such as:

- `//dbReference` - elements of type dbReference that are descendants of something:

````bash
%%bash
xmllint --xpath '//dbReference' chebi_27732_P21817_entry.xml
````

**Expected Output:** All dbReference elements from the entry.

- `/entry//dbReference` - equivalent to the previous query but specifying that the dbReference elements are descendants of the entry element:

````bash
%%bash
xmllint --xpath '/entry//dbReference' chebi_27732_P21817_entry.xml
````

**Expected Output:** All dbReference elements that are descendants of the entry element.

- `/entry/reference/citation/dbReference` - similar to the previous query but specifying the full path in the XML file, i.e. only `dbReference` elements descendants of citation, reference and entry elements:

````bash
%%bash
xmllint --xpath '/entry/reference/citation/dbReference' chebi_27732_P21817_entry.xml
````

**Expected Output:** dbReference elements following the specified path hierarchy.

- `//dbReference/*` - any child elements of a dbReference element:

````bash
%%bash
xmllint --xpath '//dbReference/*' chebi_27732_P21817_entry.xml
````

**Expected Output:** All child elements contained within dbReference elements.

- `//dbReference/property[1]` - first property element of each dbReference element:

````bash
%%bash
xmllint --xpath '//dbReference/property[1]' chebi_27732_P21817_entry.xml
````

**Expected Output:** First property element of each dbReference.

- `//dbReference/property[2]` - second property element of each dbReference element:

````bash
%%bash
xmllint --xpath '//dbReference/property[2]' chebi_27732_P21817_entry.xml
````

**Expected Output:** Second property element of each dbReference.

- `//dbReference/property[3]` - third property element of each dbReference element:

````bash
%%bash
xmllint --xpath '//dbReference/property[3]' chebi_27732_P21817_entry.xml
````

**Expected Output:** Third property element of each dbReference.

- `//dbReference/property/@type` - all type attributes of the property elements:

````bash
%%bash
xmllint --xpath '//dbReference/property/@type' chebi_27732_P21817_entry.xml
````

**Expected Output:** All type attribute values from property elements.

- `//dbReference/property[@type="protein sequence ID"]` - the previous property elements that have an attribute type equal to protein sequence ID:

````bash
%%bash
xmllint --xpath '//dbReference/property[@type="protein sequence ID"]' chebi_27732_P21817_entry.xml
````

**Expected Output:** Property elements with type attribute equal to "protein sequence ID".

- `//dbReference/property[@type="protein sequence ID"]/@value` - the string assigned to each attribute value of the previous property elements:

````bash
%%bash
xmllint --xpath '//dbReference/property[@type="protein sequence ID"]/@value' chebi_27732_P21817_entry.xml
````

**Expected Output:** Value attributes from the selected property elements.

- `/entry/sequence/text()` - the contents inside the sequence element:

````bash
%%bash
xmllint --xpath '/entry/sequence/text()' chebi_27732_P21817_entry.xml
````

**Expected Output:** The amino acid sequence contained in the sequence element.

Thus, an alternative way to extract the PubMed identifiers using `xmllint` instead of `grep`, would be something like this:

````bash
%%bash
xmllint --xpath '//dbReference[@type="PubMed"]/@id' chebi_27732_P21817_entry.xml
````

**Expected Output:** All PubMed id attributes from dbReference elements, displayed with their attribute labels.

However, the output contains all identifiers in the same line and with the id label.

Previous versions of xmllint may print all the output in the same line. In that case, we need to add an extra `tr ' ' '\n'` command to split the output in multiple lines (one line per identifier).

### Extracting XPath results

To extract the identifiers, we can use the cut command:

````bash
%%bash
xmllint --xpath '//dbReference[@type="PubMed"]/@id' chebi_27732_P21817_entry.xml | cut -d'"' -f2
````

**Expected Output:** Extracted PubMed identifiers without the attribute labels.

The cut command extracts the value inside the double quotes.
